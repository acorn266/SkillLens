```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from pathlib import Path

BASE_DIR = Path.cwd().parent

DATA_FILE = BASE_DIR / "data" / "processed" / "skill_job_cleaned.csv"

df = pd.read_csv(DATA_FILE)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df["Job_Match"].value_counts())

print("\nPercentages:")
print(df["Job_Match"].value_counts(normalize=True) * 100)
```

```python
print(df["Job_Match"].value_counts())

df["Job_Match"].value_counts().plot(
    kind="bar",
    title="Job Match Distribution"
)

plt.xlabel("Job Match")
plt.ylabel("Number of Students")
plt.show()
```

```python
print(df["Vocational_Program"].value_counts())
```

```python
df["Vocational_Program"].value_counts().plot(
    kind="bar",
    title="Students by Vocational Program"
)

plt.xlabel("Vocational Program")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.show()
```

```python
student_skill_columns = [
    "Skill_1",
    "Skill_2",
    "Skill_3",
    "Skill_4",
    "Skill_5"
]

skill_averages = df[student_skill_columns].mean()

print(skill_averages)
```

```python
skill_averages.plot(
    kind="bar",
    title="Average Student Skill Levels"
)

plt.xlabel("Skills")
plt.ylabel("Average Skill Level")
plt.ylim(0, 10)
plt.show()
```

```python
required_skill_columns = [
    "Required_Skill_1",
    "Required_Skill_2",
    "Required_Skill_3",
    "Required_Skill_4",
    "Required_Skill_5"
]

required_skill_averages = df[required_skill_columns].mean()

print(required_skill_averages)
```

```python
required_skill_averages.plot(
    kind="bar",
    title="Average Required Skill Levels"
)

plt.xlabel("Required Skills")
plt.ylabel("Average Required Level")
plt.ylim(0, 10)
plt.show()
```
